# Data Download

## Overview :-

**What this notebook does :**
- downloads the M5 Walmart dataset from Kaggle into the project's raw data folder
- checks that the files needed for the project are there
- finds out which of the downloaded files are really needed

**Approach followed :**
- the data is downloaded with [kagglehub](https://github.com/Kaggle/kagglehub), Kaggle's own Python library
- the raw files are never edited after download, every later step reads them and saves its results somewhere else

**Steps :**
- 1 . Import Required Modules
- 2 . Settings
- 3 . Kaggle Login
- 4 . Download the Data
- 5 . Check the Downloaded Files

**Reads / Writes :**
- **Reads :** [M5 Forecasting - Accuracy (Kaggle)](https://www.kaggle.com/competitions/m5-forecasting-accuracy/data)
- **Writes :** `data/M5-Walmart-Sales-Forecasting/`

# 1 . Import Required Modules :-

```python
import os                                   # import os library for working with folders and files
from getpass import getpass                 # import getpass to type the Kaggle token without showing it
import pandas as pd                         # import pandas library for reading the data

import warnings                             # import warnings to ignore future warnings
warnings.filterwarnings('ignore')

import kagglehub                            # import kagglehub library for downloading Kaggle data
```

# 2 . Settings :-

- **DATA_DIR :** folder where the raw data is saved (same as `raw_dir` in `config/stockwise.yaml`)
- **COMPETITION :** name of the competition on Kaggle, taken from the end of its web address
- **FILES :** the 5 files Kaggle gives for this competition, used to check that the download is complete

```python
DATA_DIR = '../data/M5-Walmart-Sales-Forecasting/'                       # raw data folder
COMPETITION = 'm5-forecasting-accuracy'                                   # Kaggle competition name
FILES = ['calendar.csv', 'sales_train_evaluation.csv', 'sales_train_validation.csv',
         'sample_submission.csv', 'sell_prices.csv']                      # all files of the competition
```

# 3 . Kaggle Login :-

Kaggle gives competition data only to logged in users who accepted the competition rules. One time setup :

- open the [competition page](https://www.kaggle.com/competitions/m5-forecasting-accuracy) and click **Join Competition** to accept the rules
- go to **Kaggle -> Settings -> API -> Create New Token** and copy the token
- save it in a file named `.env` in the project folder, as one line : `KAGGLE_API_TOKEN=<your token>`

The cell below reads the token from `.env`. If there is no `.env` file, a box opens to paste the token.
- `.env` is listed in `.gitignore`, so the token never goes to git
- never type the token directly inside a cell, the notebook goes to git
- if the files are already in `DATA_DIR`, no login is needed

```python
if all(os.path.exists(DATA_DIR + i) for i in FILES):                     # login is needed only for the download
    print("Data already downloaded, login not needed")
else:
    if os.path.exists('../.env'):                                         # read the token from the .env file
        for line in open('../.env'):
            if '=' in line:
                name, value = line.strip().split('=', 1)                  # KAGGLE_API_TOKEN=<token>
                os.environ[name] = value
    if 'KAGGLE_API_TOKEN' not in os.environ:                              # no .env file, so ask for the token
        os.environ['KAGGLE_API_TOKEN'] = getpass("Paste your Kaggle API token : ")
    print("Logged in as : ", kagglehub.whoami()['username'])              # check the token with Kaggle
```

# 4 . Download the Data :-

The download (about **450 MB**) runs only when the files are not already in the folder, so running the notebook again does not download the data again.

```python
os.makedirs(DATA_DIR, exist_ok=True)                                      # create the folder if it is not there

if all(os.path.exists(DATA_DIR + i) for i in FILES):                      # skip the download if the files are already there
    print("Data already downloaded in : ", DATA_DIR)
else:
    kagglehub.competition_download(COMPETITION, output_dir=DATA_DIR)      # download and unzip the competition files
    print("Data downloaded to : ", DATA_DIR)
```

# 5 . Check the Downloaded Files :-

The [Kaggle data page](https://www.kaggle.com/competitions/m5-forecasting-accuracy/data) describes the 5 files as :
- **calendar.csv :** the dates on which the products are sold, with events and SNAP days
- **sales_train_validation.csv :** daily units sold per product and store, days `d_1` to `d_1913`
- **sales_train_evaluation.csv :** the same sales with **28** more days, `d_1` to `d_1941`
- **sell_prices.csv :** the price of each product per store and week
- **sample_submission.csv :** the format for sending forecasts to the competition

The cells below check the files on disk.

```python
for i in sorted(os.listdir(DATA_DIR)):                                    # loop over the files in the folder
    if i.endswith('.csv'):
        size = os.path.getsize(DATA_DIR + i) / 1e6                        # file size in MB
        print(i, ":", round(size, 1), "MB")

missing = [i for i in FILES if not os.path.exists(DATA_DIR + i)]          # needed files that are not there
print("\nMissing files : ", missing)
```

### Insights :-
- Kaggle gives **5** csv files and none of them is missing.
- `sell_prices.csv` is the biggest file (**203.4 MB**), `calendar.csv` is the smallest (**0.1 MB**).
- There are two sales files, so we check which one to use.

## ( i ) . Which Sales File to Use :-

There are two sales files. First we look at the first rows of both, then we compare them to decide which one we need.

```python
val = pd.read_csv(DATA_DIR + 'sales_train_validation.csv', nrows=5)          # read only the first 5 rows
val
```

```python
eva = pd.read_csv(DATA_DIR + 'sales_train_evaluation.csv', nrows=5)          # read only the first 5 rows
eva
```

### Insights :-
- Each row is **one product in one store** (`item_id` and `store_id`), with its department, category and state.
- Each day is a **column** : `d_1`, `d_2`, ... up to `d_1913` in the validation file and up to `d_1941` in the evaluation file.
- The validation file has **1,919** columns and the evaluation file **1,947**, the first **6** describe the product and the rest are days.
- So we compare the two files on their products (rows) and their days (columns).

```python
val_days = [i for i in val.columns if i.startswith('d_')]                   # day columns of validation file
eva_days = [i for i in eva.columns if i.startswith('d_')]                   # day columns of evaluation file

## products of all rows, only the 2 product columns are read
val_products = pd.read_csv(DATA_DIR + 'sales_train_validation.csv', usecols=['item_id', 'store_id'])
eva_products = pd.read_csv(DATA_DIR + 'sales_train_evaluation.csv', usecols=['item_id', 'store_id'])

## sales of the common days d_1 to d_1913 from both files (int16 keeps the memory small)
val_sales = pd.read_csv(DATA_DIR + 'sales_train_validation.csv', usecols=val_days, dtype='int16')
eva_sales = pd.read_csv(DATA_DIR + 'sales_train_evaluation.csv', usecols=val_days, dtype='int16')

print("Products in validation file   : ", val_products.shape[0])
print("Products in evaluation file   : ", eva_products.shape[0])
print("Same products in both files   : ", val_products.equals(eva_products))
print("Days in validation file       : ", len(val_days), "(", val_days[0], "to", val_days[-1], ")")
print("Days in evaluation file       : ", len(eva_days), "(", eva_days[0], "to", eva_days[-1], ")")
print("Same sales on the common days : ", val_sales.equals(eva_sales))
```

### Insights :-
- Both files have the same **30,490** products.
- The evaluation file has **1,941** days and the validation file **1,913**, so the evaluation file has **28** more days at the end.
- The sales of the common days `d_1` to `d_1913` are exactly the same in both files.
- So the evaluation file has everything the validation file has, and we use only `sales_train_evaluation.csv`. `sample_submission.csv` is only the Kaggle submission format, so it is not needed either.
