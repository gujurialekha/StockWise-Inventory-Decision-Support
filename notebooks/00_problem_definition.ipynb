# Project Overview :-

**StockWise : Machine Learning-Based Sales Forecasting and Inventory Decision Support**

StockWise forecasts the weekly sales of selected products in a retail store and uses the forecast to support stock planning. Machine learning models are compared with simple methods, and an interactive dashboard shows sales trends, forecasts and model performance.

- **Who uses it :** a retail analyst or store planner who orders stock for the store every week.
- **Decision :** how many units of each product to order this week.
- **Cost of a wrong decision :** too little stock loses sales, too much stock keeps money and shelf space stuck.
- **What the project gives :** a sales forecast for the next 4 weeks, the products likely to sell more or less than usual, and a suggested order quantity, with the data period, forecast date and stock assumptions shown, so the planner can check how the number was made.
- **Who decides :** the planner. The project only suggests the order, it does not place it.

# 1 . Problem Statement :-

- A store has to order stock **before** it knows how much will sell in the coming weeks.
- A sales forecast alone does not give the order quantity. We also need the stock on hand, the orders already on the way and the supplier lead time.
- If the forecast is never tested, an order calculator built on it gives false confidence. A complex model is not always better than a simple one, so this has to be tested, not assumed.
- So in this project we forecast the **next 4 weeks of sales** for selected products of one store, test the forecasts fairly against simple methods, and turn them into order quantities for a given stock situation.

# 2 . Objectives :-

- **( i ) .** Build a clean weekly sales table for the selected products, and check that no sales are lost when daily sales are added up into weeks.
- **( ii ) .** Forecast the sales of the next 4 weeks and compare machine learning models with simple methods, on the same weeks, products and metrics.
- **( iii ) .** Measure how close the forecasts are to the actual sales on weeks the models have never seen.
- **( iv ) .** Turn the forecast into an order quantity for a given stock situation, and test a simple ordering rule on past weeks.
- **( v ) .** Show sales trends, forecasts and model performance in an interactive dashboard.

# 3 . Approach :-

- **Type :** regression on time series (forecasting).
- **What we predict :** units sold of a product in a Monday-Sunday week. This is the sales that happened, not the true demand, because sales stop when a product is out of stock.
- **How far ahead :** **1 to 4** weeks.
- **When we predict :** at the end of every Sunday, using only the data known up to that Sunday. The calendar of the coming weeks (holidays, events) is known in advance, so it can be used.
- **Main metric :** macro RMSSE. For each product, the forecast error is compared with how much its sales normally change week to week, then the result is averaged over all products with equal weight. Weekly sales of a single product can be **zero**, so MAPE is not used (it divides by the actual value).
- **Other metrics :** MASE, MAE, RMSE, WAPE and bias.
- **Baselines :** every model is compared with **last week naive** (same as last week) and **seasonal naive** (same week last year). If a simple method wins, that is reported as it is.

# 4 . Data Requirements :-

The problem needs data with :

- **units sold per product per day or week, in one store :** this is what we forecast
- **at least 3 years of history :** to compare with the same week last year and to learn yearly patterns
- **a calendar of holidays and events :** these are known in advance and change sales
- **selling prices per product :** to check whether price helps the forecast
- **many products of different types :** so the results do not depend on one product
- **public and free to download :** so the work can be repeated by anyone

# 5 . Dataset Selection :-

**Selected : M5 Walmart dataset** - [M5 Forecasting - Accuracy (Kaggle)](https://www.kaggle.com/competitions/m5-forecasting-accuracy/data)
- meets every requirement above : daily unit sales per product per store for more than 5 years, a calendar with holidays, events and SNAP days, and weekly sell prices, all in one place

**Other options checked :**
- **Online Retail II** - [UCI Machine Learning Repository](https://archive.ics.uci.edu/dataset/502/online+retail+ii) : transaction level data covering about 2 years only, needs heavy cleaning and has no event calendar.
- **Store Item Demand** - [Store Item Demand Forecasting Challenge (Kaggle)](https://www.kaggle.com/competitions/demand-forecasting-kernels-only/data) : simple daily sales, but no prices and no events.

**What M5 does not have :**
- stock on hand, lead time and costs, so in the order calculator these are given as scenario values
- stockout flags, so a zero-sales week can be low demand or an empty shelf and is not treated as a stockout

# 6 . Scope & Limitations :-

**In scope :**
- one Walmart store, **CA_1** (California)
- two categories, **FOODS** and **HOUSEHOLD**, about **100** products that sell regularly
- weekly data from **2013** to **May 2016**

**Out of scope :**
- placing orders automatically
- other stores, new products and products that rarely sell
- saying that a price change **caused** a sales change

**Limitations :**
- results for one store and regular products may not hold for other stores or for today's market
- the order simulation uses past sales as demand, so it shows a scenario, not real savings
